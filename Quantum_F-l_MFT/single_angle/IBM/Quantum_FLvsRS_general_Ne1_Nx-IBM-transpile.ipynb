# Quantum_FL_RS_general_Ne_Nx

Clean generalized notebook for comparing the many-body Qiskit evolution with the Raffelt--Sigl mean-field evolution for

\[
N_e\;\nu_e + N_x\;\nu_x(\alpha),
\qquad
\nu_x(\alpha)=\cos\alpha\,\nu_e+\sin\alpha\,\nu_\mu .
\]

This is the generalized version of the older fixed \(1\nu_e+15\nu_x\) notebook.

Default choice:

\[
N_e=1,\qquad N_x=15.
\]

To check \(N_e=2\) later, change only `Ne` in Cell 1.

```python
# =====================================================
# IBM observable construction
# =====================================================

def make_ibm_z_observable(N, qubit):
    """
    Construct the single-qubit Z observable acting on `qubit`.

    Qiskit's Pauli strings are written with qubit N-1 on the left
    and qubit 0 on the right.
    """
    pauli = ["I"] * N
    pauli[N - 1 - qubit] = "Z"

    return SparsePauliOp.from_list([
        ("".join(pauli), 1.0)
    ])
```

```python
def heisenberg_interaction_swap(
    qc,
    q1,
    q2,
    phi
):
    """
    H-SWAP = exp[-i phi/2 H_H] * SWAP
    up to a global phase,

    where H_H = XX + YY + ZZ.

    SWAP = exp[i*pi/4] *
           exp[-i*pi/4 (XX+YY+ZZ)]

    Therefore:

    H-SWAP ~
    exp[-i (phi/2 + pi/4)(XX+YY+ZZ)]

    and in Qiskit rotation-angle convention:

        theta = phi + pi/2
    """

    theta = phi + np.pi / 2

    qc.rxx(theta, q1, q2)
    qc.ryy(theta, q1, q2)
    qc.rzz(theta, q1, q2)
```

```python
# ============================================================
# Find a 10-qubit path on Kingston
# The two endpoints will host the two nu_e qubits.
# ============================================================

# ============================================================
# 1. Select the Kingston path
# ============================================================

def find_kingston_path_2e8x(backend, n_nodes=10, top_k=30):

    cmap = backend.coupling_map

    adjacency = {
        q: set()
        for q in range(backend.num_qubits)
    }

    for a, b in cmap:
        adjacency[a].add(b)
        adjacency[b].add(a)

    degree = {
        q: len(adjacency[q])
        for q in adjacency
    }

    candidates = sorted(
        degree,
        key=lambda q: degree[q],
        reverse=True
    )[:top_k]

    best_path = None
    best_score = -np.inf

    def dfs(path):

        nonlocal best_path, best_score

        if len(path) == n_nodes:

            endpoint_score = (
                degree[path[0]]
                + degree[path[-1]]
            )

            internal_score = sum(
                degree[q]
                for q in path[1:-1]
            )

            score = (
                100.0 * endpoint_score
                + 10.0 * internal_score
            )

            if score > best_score:
                best_score = score
                best_path = path.copy()

            return

        current = path[-1]

        neighbors = sorted(
            adjacency[current],
            key=lambda q: degree[q],
            reverse=True
        )

        for q in neighbors:

            if q in path:
                continue

            dfs(path + [q])

    for start in candidates:
        dfs([start])

    if best_path is None:
        raise RuntimeError(
            "Could not find a 10-qubit path on Kingston."
        )

    print("\nSelected Kingston physical path:")
    print(best_path)

    print("\nPhysical connectivity:")
    for q in best_path:
        print(
            f"q{q}: degree = {degree[q]}"
        )

    print(
        "\nEndpoint connectivities: "
        f"{degree[best_path[0]]} and "
        f"{degree[best_path[-1]]}"
    )

    return best_path


# ============================================================
# Kingston layout for:
#     q0,q1     -> nu_e
#     q2...q9   -> nu_x
#
# Physical path:
#     nu_e - nu_x - ... - nu_x - nu_e
# ============================================================

def get_kingston_layout_2e8x(backend):

    path = find_kingston_path_2e8x(
        backend,
        n_nodes=10
    )

    # Logical -> physical mapping
    #
    # logical q0 -> path[0]       (nu_e)
    # logical q1 -> path[9]       (nu_e)
    # logical q2 -> path[1]       (nu_x)
    # ...
    # logical q9 -> path[8]       (nu_x)

    initial_layout = (
        [path[0], path[-1]]
        + path[1:-1]
    )

    print("\nLogical -> physical layout:")
    for logical, physical in enumerate(initial_layout):
        species = "nu_e" if logical < 2 else "nu_x"
        print(
            f"logical q{logical} ({species})"
            f" -> physical q{physical}"
        )

    return path, initial_layout
```

```python
# =========================================
# CELL 3: Qiskit evolution building blocks
# =========================================

def add_vacuum_evolution(qc, n, bx, by, bz, N):
    """
    Add n first-order vacuum steps to the circuit.

    This function is retained for completeness, but use_vacuum=False by default
    to match the old e1_x15 comparison.
    """
    for _ in range(n):
        for q in range(N):
            qc.rx(2.0 * bx, q)
            qc.ry(2.0 * by, q)
            qc.rz(2.0 * bz, q)


def add_interaction_evolution(qc, n, J, N, dt):
    """
    Add n first-order interaction steps to the circuit.

    For each pair (i,j), apply approximately

        exp[-i Jij dt (XX + YY + ZZ)]

    using

        RXX(2 Jij dt) RYY(2 Jij dt) RZZ(2 Jij dt).
    """
    for _ in range(n):
        for i in range(N):
            for j in range(i + 1, N):
                phi = 1.0 * J* dt
                qc.rxx(phi, i, j)
                qc.ryy(phi, i, j)
                qc.rzz(phi, i, j)




# ============================================================
# IBM Kingston optimized nu_e-nu_x interaction
# ============================================================


def add_interaction_evolution_ibm(
    qc, n, J, Ne, Nx, dt
):
    """
    Add only nu_e-nu_x interactions.

    q[0 ... Ne-1]       = nu_e
    q[Ne ... N-1]       = nu_x
    """

    N = Ne + Nx

    if qc.num_qubits != N:
        raise ValueError(
            f"qc has {qc.num_qubits} qubits, "
            f"but Ne + Nx = {N}"
        )

    phi = J * dt

    for _ in range(n):

        for i in range(Ne):

            for j in range(Ne, N):

                qc.rxx(phi, i, j)
                qc.ryy(phi, i, j)
                qc.rzz(phi, i, j)


# ============================================================
# Kingston-specific interaction network
# Ne = 2, Nx = 8
#
# The two nu_e states sweep through all eight nu_x states.
# Every nu_e-nu_x pair interacts exactly once per Trotter step.
#
# There are:
#   16 interaction+SWAP operations
#   1 nu_e-nu_e SWAP
# per Trotter step.
# ============================================================

def add_interaction_evolution_kingston_2e8x(
    qc,
    n,
    J,
    dt,
    path
):
    """
    Kingston hardware-aware first-order Trotter evolution
    for Ne=2, Nx=8.

    Physical path:

        e - x - x - x - x - x - x - x - x - e

    The gates are always applied between ADJACENT PHYSICAL
    QUBITS in `path`.

    Each H-SWAP implements

        exp[-i phi/2 (XX+YY+ZZ)] * SWAP

    up to a global phase, with

        phi = J*dt.

    The Python list `order` tracks which logical flavor state
    occupies each physical position.
    """

    phi = J * dt

    for step in range(n):

        # -------------------------------------------------
        # Current logical-state ordering along the physical
        # Kingston path.
        #
        # Initially:
        #
        #   [e, x, x, x, x, x, x, x, x, e]
        # -------------------------------------------------

        order = list(range(len(path)))

        # -------------------------------------------------
        # LEFT ELECTRON SWEEP
        #
        # Electron initially at position 0.
        #
        # It interacts with x states at positions
        # 1,2,...,8 while moving through the chain.
        #
        # Every gate is between adjacent physical qubits.
        # -------------------------------------------------

        for k in range(8):

            q1 = path[k]
            q2 = path[k + 1]

            heisenberg_interaction_swap(
                qc,
                q1,
                q2,
                phi
            )

            # Logical states exchange positions because
            # H-SWAP contains a SWAP.
            order[k], order[k + 1] = (
                order[k + 1],
                order[k]
            )

        # -------------------------------------------------
        # At this point the first electron has moved to
        # position 8.
        #
        # The second electron is at position 9.
        #
        # Swap the two electron states so that the second
        # electron can sweep back through the x states.
        # -------------------------------------------------

        q1 = path[8]
        q2 = path[9]

        qc.swap(q1, q2)

        order[8], order[9] = (
            order[9],
            order[8]
        )

        # -------------------------------------------------
        # RIGHT ELECTRON SWEEP
        #
        # The second electron now moves from position 8
        # back through positions 7,...,0.
        #
        # Again, every interaction is on an adjacent
        # physical Kingston edge.
        # -------------------------------------------------

        for k in range(8, 0, -1):

            q1 = path[k]
            q2 = path[k - 1]

            heisenberg_interaction_swap(
                qc,
                q1,
                q2,
                phi
            )

            order[k], order[k - 1] = (
                order[k - 1],
                order[k]
            )
```

```python
# ============================================
# CELL 4: Raffelt--Sigl mean-field solver
# ============================================

# Pauli matrices
sigma_0 = np.array([[1, 0],  [0,  1]], dtype=complex)
sigma_1 = np.array([[0, 1],  [1,  0]], dtype=complex)
sigma_2 = np.array([[0, -1j], [1j, 0]], dtype=complex)
sigma_3 = np.array([[1, 0],  [0, -1]], dtype=complex)


def P_osc_RS(t_table, theta, omega, lam, J, initial_flavors=None, alpha=None):
    """
    Raffelt--Sigl two-flavor polarization-vector evolution.

    Parameters
    ----------
    t_table : array
        Sampled times.
    theta : float
        Vacuum mixing angle.
    omega : array
        Vacuum frequencies for all modes.
    lam : float
        Matter strength.
    J : array
        Coupling matrix.
    initial_flavors : array
        Entries may be:
            'e'  : |nu_e>
            'x'  : |nu_mu>
            'mu' : |nu_mu>
            'a'  : cos(alpha)|nu_e> + sin(alpha)|nu_mu>
    alpha : float
        Superposition angle for every mode labeled 'a'.
    """
    n_modes = omega.size

    u = np.array([[np.cos(theta), np.sin(theta)],
                  [-np.sin(theta), np.cos(theta)]])
    b = 0.5 * np.diag([-1, 1])
    b = u @ b @ u.T
    l = np.diag([1, 0])

    Ee = np.array([0.0, 0.0,  1.0])
    Emu = np.array([0.0, 0.0, -1.0])

    B = np.real(np.array([np.trace(b @ sigma_1),
                          np.trace(b @ sigma_2),
                          np.trace(b @ sigma_3)]))
    L = np.real(np.array([np.trace(l @ sigma_1),
                          np.trace(l @ sigma_2),
                          np.trace(l @ sigma_3)]))

    ini_state = np.where(omega > 0, Ee[:, None], Emu[:, None])

    if initial_flavors is not None:
        initial_flavors = np.asarray(initial_flavors)

        for f in initial_flavors:
            if f not in ["e", "x", "mu", "a"]:
                raise ValueError(f"Unknown flavor label: {f}")
            if f == "a" and alpha is None:
                raise ValueError("Flavor label 'a' requires alpha.")

        ini_state[:, initial_flavors == "e"] = Ee[:, None]
        ini_state[:, initial_flavors == "x"] = Emu[:, None]
        ini_state[:, initial_flavors == "mu"] = Emu[:, None]
        ini_state[:, initial_flavors == "a"] = np.array(
            [np.sin(2.0 * alpha), 0.0, np.cos(2.0 * alpha)]
        )[:, None]

    def rhs(t, P_flat):
        """
        dP_k/dt = H_k x P_k,
        H_k = omega_k B + lambda L + sum_j J_kj P_j.
        """
        res = np.zeros(3 * n_modes)
        PP = np.reshape(P_flat, (n_modes, 3)).T
        
        for k in range(n_modes):
            H_k = omega[k] * B + lam * L + J * np.sum(PP, axis=1)
            res[3*k:3*k+3] = np.cross(H_k, PP[:, k])

        return res
    
    return integ.solve_ivp(
        rhs,
        (t_table[0], t_table[-1]),
        ini_state.T.flatten(),
        t_eval=t_table,
    )
```

```python
# =====================================================
# CELL 5: Many-body Qiskit solver for Ne + Nx(alpha)
# =====================================================

def prepare_initial_circuit(Ne, Nx, alpha):
    """
    Prepare the initial many-body state:

        first Ne qubits: |nu_e> = |0>
        next  Nx qubits: |nu_x(alpha)> = cos(alpha)|0> + sin(alpha)|1>

    RY(2 alpha)|0> = cos(alpha)|0> + sin(alpha)|1>.

    For alpha = pi/2, the Nx group is exactly |1>, reproducing
    the old 1 nu_e + 15 nu_mu setup when Ne=1, Nx=15.
    """
    N = Ne + Nx
    qc0 = QuantumCircuit(N, N)

    # First Ne qubits are |nu_e> = |0>, so no gate is needed.
    for q in range(Ne, N):
        qc0.ry(2.0 * alpha, q)

    return qc0


def run_many_body_qiskit(alpha, Ne, Nx, J, times, dt, shots, backend,
                         use_vacuum=False, bx=0.0, by=0.0, bz=0.0,
                         verbose=True):
    """
    Run the many-body Qiskit simulation for a given alpha.

    Output
    ------
    P_bit1[q, k] :
        Probability that qubit q is measured as |1> at time times[k].

    For the first Ne qubits, which start as |nu_e>=|0>,
        P_bit1 = P(nu_e -> nu_mu).

    The plotted observable is the Ne-average:
        <P(nu_e -> nu_mu)> over the initially electron-flavor group.
    """
    N = Ne + Nx
    qc0 = prepare_initial_circuit(Ne, Nx, alpha)

    P_bit1 = [[] for _ in range(N)]

    for t in times:
        n = int(round(t / dt))
        qc = qc0.copy()

        #if use_vacuum:
            #add_vacuum_evolution(qc, n=n, bx=bx, by=by, bz=bz, N=N)

        add_interaction_evolution(qc, n=n, J=J, N=N, dt=dt)

        qc.measure(range(N), range(N))

        tqc = transpile(qc, backend, optimization_level=0)
        counts = backend.run(tqc, shots=shots).result().get_counts()

        p1_t = [0.0] * N

        for bitstring, c in counts.items():
            prob = c / shots

            for q in range(N):
                # Qiskit displays bitstrings with the highest classical bit first.
                bit_q = int(bitstring[-1 - q])
                if bit_q == 1:
                    p1_t[q] += prob

        for q in range(N):
            P_bit1[q].append(p1_t[q])

        if verbose:
            print(
                f"alpha={alpha:.6f}, t={t:8.3f}: "
                f"<P_e_to_mu>_MB = {np.mean(p1_t[:Ne]):.5f}"
            )

    P_bit1 = np.array(P_bit1)

    return {
        "P_bit1": P_bit1,
        "P_e_to_mu_avg": np.mean(P_bit1[:Ne, :], axis=0),
    }
```

```python
# =====================================================
# CELL 6: RS solver and combined alpha-case runner
# =====================================================

def run_rs_mean_field(alpha, Ne, Nx, J, times,
                      use_vacuum=False, use_matter=False):
    """
    Run the Raffelt--Sigl mean-field evolution for the same initial state.

    Default:
        omega = 0,
        lambda = 0,

    so this solves the self-interaction-only mean-field problem, matching
    the old e1_x15 comparison.
    """
    N = Ne + Nx

    if use_vacuum:
        omega = np.ones(N) * omega1
        theta_rs = theta_vac
    else:
        omega = np.zeros(N)
        theta_rs = 0.0

    lam = 1.0 if use_matter else 0.0

    initial_flavors = np.array(["e"] * Ne + ["a"] * Nx)

    sol = P_osc_RS(
        times,
        theta_rs,
        omega,
        lam,
        J,
        initial_flavors=initial_flavors,
        alpha=alpha,
    )

    P_all = sol.y
    P = P_all.reshape(N, 3, -1)
    Pz = P[:, 2, :]

    # P(nu_mu) = (1 - Pz)/2 for each mode.
    P_mu = 0.5 * (1.0 - Pz)

    return {
        "sol": sol,
        "P": P,
        "Pz": Pz,
        "P_mu": P_mu,
        "P_e_to_mu_avg": np.mean(P_mu[:Ne, :], axis=0),
    }


def run_alpha_case(alpha, verbose=True):
    """
    Run both many-body Qiskit and RS mean-field for one alpha.
    """
    N = Ne + Nx
    J = 1

    mb = run_many_body_qiskit(
        alpha=alpha,
        Ne=Ne,
        Nx=Nx,
        J=J,
        times=times,
        dt=dt,
        shots=shots,
        backend=backend,
        use_vacuum=use_vacuum,
        bx=bx,
        by=by,
        bz=bz,
        verbose=verbose,
    )

    rs = run_rs_mean_field(
        alpha=alpha,
        Ne=Ne,
        Nx=Nx,
        J=J,
        times=times,
        use_vacuum=use_vacuum,
        use_matter=use_matter,
    )

    return {
        "alpha": alpha,
        "Ne": Ne,
        "Nx": Nx,
        "many_body": mb,
        "rs": rs,
    }
```

```python
def build_ibm_circuit(
    alpha,
    Ne,
    Nx,
    J,
    t,
    dt,
    backend
):

    if Ne != 2 or Nx != 8:
        raise ValueError(
            "This Kingston-specific circuit requires "
            "Ne=2 and Nx=8."
        )

    N = Ne + Nx

    if backend.num_qubits < 10:
        raise ValueError(
            "Backend does not have enough physical qubits."
        )

    # --------------------------------------------------------
    # Find the Kingston path
    # --------------------------------------------------------

    path = find_kingston_path_2e8x(
        backend,
        n_nodes=N
    )

    # --------------------------------------------------------
    # Build directly on all physical Kingston qubits
    # --------------------------------------------------------

    qc = QuantumCircuit(
        backend.num_qubits
    )

    # --------------------------------------------------------
    # Initial flavor state
    #
    # path[0]     = nu_e = |0>
    # path[1:9]   = nu_x
    # path[9]     = nu_e = |0>
    #
    # nu_x = cos(alpha)|0> + sin(alpha)|1>
    #      = RY(2 alpha)|0>
    # --------------------------------------------------------

    # The two nu_e states are already |0>.
    #
    # Prepare the eight nu_x states.
    for physical_q in path[1:9]:

        qc.ry(
            2 * alpha,
            physical_q
        )

    # --------------------------------------------------------
    # Trotter evolution
    # --------------------------------------------------------

    n = int(round(t / dt))

    add_interaction_evolution_kingston_2e8x(
        qc,
        n=n,
        J=J,
        dt=dt,
        path=path
    )

    return qc, path

# ============================================================
# 5. Transpile
#
# NO initial_layout.
# NO SABRE.
# NO routing.
#
# The circuit has already been constructed directly on
# connected Kingston physical qubits.
# ============================================================

def transpile_ibm_circuit(
    qc,
    backend
):

    tqc = transpile(
        qc,
        backend=backend,
        optimization_level=3,
        routing_method="none",
        seed_transpiler=42,
    )

    return tqc
```

```python
from qiskit import QuantumCircuit, ClassicalRegister, transpile


def run_many_body_ibm(
    alpha,
    Ne,
    Nx,
    J,
    times,
    dt,
    sampler,
    backend,
    shots=shots,
    verbose=True,
):

    N = Ne + Nx

    # =====================================================
    # This implementation currently uses the direct
    # Kingston physical-qubit construction for Ne=2, Nx=8.
    # =====================================================

    if not (Ne == 2 and Nx == 8):
        raise ValueError(
            "The direct Kingston SWAP-network implementation "
            "currently requires Ne=2 and Nx=8."
        )

    # =====================================================
    # Initialize storage
    # =====================================================

    P_bit1 = [[] for _ in range(N)]
    P_bit1_std = [[] for _ in range(N)]

    P_e_to_mu_avg = []
    P_e_to_mu_std = []

    # Store all transpiled circuits
    transpiled_circuits = []

    # Store circuit depths
    circuit_depths = []

    # =====================================================
    # IBM run information
    # =====================================================

    if verbose:
        print()
        print("=" * 60)
        print("Preparing IBM hardware run")
        print("=" * 60)
        print(f"alpha       = {alpha:.6f}")
        print(f"Ne          = {Ne}")
        print(f"Nx          = {Nx}")
        print(f"N           = {N}")
        print(f"J           = {J}")
        print(f"dt          = {dt}")
        print(f"shots       = {shots}")
        print(f"time points = {len(times)}")
        print(f"backend     = {backend.name}")
        print("=" * 60)

    # =====================================================
    # 1. Build and transpile ALL time points
    # =====================================================

    if verbose:
        print()
        print("Preparing all circuits...")
        print()

    for t in times:

        # -------------------------------------------------
        # Build direct physical-qubit circuit
        # -------------------------------------------------

        qc, path = build_ibm_circuit(
            alpha=alpha,
            Ne=Ne,
            Nx=Nx,
            J=J,
            t=t,
            dt=dt,
            backend=backend
        )

        if verbose:
            print()
            print("-" * 60)
            print(f"t = {t:.3f}")
            print(f"Total physical qubits: {qc.num_qubits}")
            print(f"Active Kingston path: {path}")

        # -------------------------------------------------
        # Diagnostic: show two-qubit gates
        # -------------------------------------------------

        if verbose:
            print("\nTwo-qubit gates in physical circuit:")

            for inst in qc.data:

                qargs = [
                    qc.find_bit(q).index
                    for q in inst.qubits
                ]

                if len(qargs) == 2:
                    print(
                        f"{inst.operation.name:25s} "
                        f"q{qargs[0]} - q{qargs[1]}"
                    )

        # -------------------------------------------------
        # Add classical register
        #
        # We only measure the N=10 active qubits in `path`.
        # The circuit itself has 156 physical qubits.
        # -------------------------------------------------

        creg = ClassicalRegister(N, "c")
        qc.add_register(creg)

        # -------------------------------------------------
        # Measure active physical qubits in PATH ORDER
        #
        # c[0] <- path[0]
        # c[1] <- path[1]
        # ...
        # c[9] <- path[9]
        # -------------------------------------------------

        for k, physical_q in enumerate(path):
            qc.measure(physical_q, creg[k])

        # -------------------------------------------------
        # Transpile
        #
        # The circuit is already constructed directly on
        # physical Kingston qubits.
        #
        # Identity layout keeps circuit q[i] on physical
        # Kingston qubit i.
        # -------------------------------------------------

        tqc = transpile_ibm_circuit(
            qc,
            backend=backend
        )

        # -------------------------------------------------
        # Store circuit and depth
        # -------------------------------------------------

        transpiled_circuits.append(tqc)

        depth = tqc.depth()
        circuit_depths.append(depth)

        if verbose:
            print(
                f"Transpiled depth = {depth}"
            )

    # =====================================================
    # 2. Print ALL circuit depths before submitting
    # =====================================================

    if verbose:

        print()
        print("=" * 60)
        print("Prepared IBM circuits")
        print("=" * 60)

        for t, depth in zip(times, circuit_depths):

            n_steps = int(round(t / dt))

            print(
                f"t={t:8.3f} | "
                f"steps={n_steps:5d} | "
                f"depth={depth:6d}"
            )

        print("=" * 60)

    # =====================================================
    # 3. Submit ALL time points in ONE SamplerV2 job
    # =====================================================

    job = sampler.run(
        transpiled_circuits,
        shots=shots
    )

    # =====================================================
    # 4. Wait for the single job
    # =====================================================

    results = job.result()

    if verbose:
        print("IBM job completed.")

    # =====================================================
    # 5. Extract result for EVERY time point
    # =====================================================

    for i, (t, pub_result) in enumerate(
        zip(times, results)
    ):

        # -------------------------------------------------
        # Extract measurement data
        # -------------------------------------------------

        bit_array = pub_result.data.c

        counts = bit_array.get_counts()

        if verbose:
            print()
            print(f"Raw counts for t={t:.3f}:")
            print(counts)
            print(f"Total shots: {sum(counts.values())}")

        # -------------------------------------------------
        # Calculate P(bit=1) for each ACTIVE qubit
        #
        # q = 0,...,9 refers to position in `path`,
        # NOT physical Kingston qubit number.
        # -------------------------------------------------

        p1_t = [0.0] * N
        p1_std_t = [0.0] * N

        for q in range(N):

            n1 = 0

            for bitstring, c in counts.items():

                # Qiskit classical-bit ordering:
                # c[0] is the rightmost bit.
                bit_q = int(
                    bitstring[-1 - q]
                )

                if bit_q == 1:
                    n1 += c

            # Probability of |1>
            p = n1 / shots

            # Binomial statistical uncertainty
            sigma = np.sqrt(
                p * (1.0 - p) / shots
            )

            p1_t[q] = p
            p1_std_t[q] = sigma

        # -------------------------------------------------
        # Store individual active-qubit results
        # -------------------------------------------------

        for q in range(N):

            P_bit1[q].append(
                p1_t[q]
            )

            P_bit1_std[q].append(
                p1_std_t[q]
            )

        # -------------------------------------------------
        # Average over the TWO electron-flavor states
        #
        # In the SWAP network the electron states are
        # located at the two endpoints of the path.
        #
        # c[0]  -> path[0]  (electron)
        # c[9]  -> path[9]  (electron)
        # -------------------------------------------------

        electron_indices = [
            0,
            N - 1
        ]

        P_avg = np.mean(
            [
                p1_t[q]
                for q in electron_indices
            ]
        )

        P_avg_std = (
            np.sqrt(
                np.sum(
                    np.asarray(
                        [
                            p1_std_t[q]
                            for q in electron_indices
                        ]
                    ) ** 2
                )
            )
            / Ne
        )

        P_e_to_mu_avg.append(
            P_avg
        )

        P_e_to_mu_std.append(
            P_avg_std
        )

        # -------------------------------------------------
        # Print result
        # -------------------------------------------------

        if verbose:

            print(
                f"t={t:8.3f}: "
                f"P_e_to_mu="
                f"{P_avg:.6f} ± "
                f"{P_avg_std:.6f}"
            )

    # =====================================================
    # 6. Convert to numpy arrays
    # =====================================================

    P_bit1 = np.asarray(P_bit1)
    P_bit1_std = np.asarray(P_bit1_std)

    P_e_to_mu_avg = np.asarray(
        P_e_to_mu_avg
    )

    P_e_to_mu_std = np.asarray(
        P_e_to_mu_std
    )

    circuit_depths = np.asarray(
        circuit_depths
    )

    # =====================================================
    # 7. Final summary
    # =====================================================

    if verbose:

        print()
        print("=" * 60)
        print("IBM run completed")
        print("=" * 60)

        for t, p, sp in zip(
            times,
            P_e_to_mu_avg,
            P_e_to_mu_std,
        ):

            print(
                f"t={t:8.3f} : "
                f"P_e_to_mu="
                f"{p:.6f} ± {sp:.6f}"
            )

        print("=" * 60)

    # =====================================================
    # Return
    # =====================================================

    return {
        "times": np.asarray(times),

        "P_bit1": P_bit1,
        "P_bit1_std": P_bit1_std,

        "P_e_to_mu_avg": P_e_to_mu_avg,
        "P_e_to_mu_std": P_e_to_mu_std,

        "circuit_depths": circuit_depths,
    }
```

## Figures for different \(\alpha\)

Each cell below runs one \(\alpha\) case and produces one figure.

The plotted observable is the average conversion probability of the initially electron-flavor group:

\[
\left\langle P(\nu_e\rightarrow \nu_\mu)\right\rangle_{N_e}.
\]

```python
import numpy as np
from scipy.special import gammaln
from sympy.physics.wigner import clebsch_gordan
from sympy import S as sympy_S


def cg_coefficient(j1, m1, j2, m2, J, M):
    """
    Clebsch-Gordan coefficient

        <j1,m1; j2,m2 | J,M>

    using the SymPy convention.
    """
    return float(
        clebsch_gordan(
            sympy_S(j1),
            sympy_S(j2),
            sympy_S(J),
            sympy_S(m1),
            sympy_S(m2),
            sympy_S(M)
        )
    )


def collective_oscillation_probability(N1, N2, alpha, t, lam=1.0):
    """
    Collective nu_e -> nu_mu oscillation probability for a two-beam system.

    Parameters
    ----------
    N1 : int
        Number of initial nu_e neutrinos in beam 1.

    N2 : int
        Number of neutrinos in beam 2.

    alpha : float
        Mixing angle appearing in
            |x> = cos(alpha)|e> + sin(alpha)|mu>.

    t : float or array-like
        Evolution time.

    lam : float, optional
        Interaction strength lambda.

    Returns
    -------
    P_emu : float or ndarray
        Collective nu_e -> nu_mu conversion probability
        for a neutrino in beam 1.
    """

    # ------------------------------------------------------------
    # Input checks
    # ------------------------------------------------------------

    if N1 < 1 or N2 < 1:
        raise ValueError("N1 and N2 must be positive integers.")

    if not isinstance(N1, (int, np.integer)):
        raise TypeError("N1 must be an integer.")

    if not isinstance(N2, (int, np.integer)):
        raise TypeError("N2 must be an integer.")

    # ------------------------------------------------------------
    # Spin quantum numbers of the two beams
    # ------------------------------------------------------------

    j1 = N1 / 2
    j2 = N2 / 2

    # Maximum total spin
    Jmax = j1 + j2

    # Make t an array so that the same code works for
    # scalar or array-valued t
    t = np.asarray(t, dtype=float)

    # Total expected number of muon neutrinos in beam 1
    Nmu = np.zeros_like(t, dtype=float)

    c = np.cos(alpha)
    s = np.sin(alpha)

    # ============================================================
    # SPECIAL CASE: alpha = pi/2
    #
    # |x> = |mu>
    #
    # Therefore only i = N2 contributes.
    # ============================================================

    if np.isclose(alpha, np.pi / 2):

        i = N2

        # Total magnetic quantum number
        M = Jmax - i

        # Allowed minimum total spin
        Smin = max(M, abs(j1 - j2))

        # --------------------------------------------------------
        # j = number of nu_e -> nu_mu conversions in beam 1
        # --------------------------------------------------------

        for j in range(1, min(i, N1) + 1):

            # Initial product state:
            #
            # |j1, j1> |j2, j2-i>
            #
            m1_initial = j1
            m2_initial = j2 - i

            # Final product state:
            #
            # |j1, j1-j> |j2, j2-i+j>
            #
            m1_final = j1 - j
            m2_final = j2 - i + j

            # Real and imaginary parts of transition amplitude
            real_part = np.zeros_like(t, dtype=float)
            imag_part = np.zeros_like(t, dtype=float)

            # ----------------------------------------------------
            # Sum over total spin S
            # ----------------------------------------------------

            S_values = np.arange(Smin, Jmax + 1, 1)

            for Stotal in S_values:

                # CG coefficient of the initial state
                C_initial = cg_coefficient(
                    j1, m1_initial,
                    j2, m2_initial,
                    Stotal, M
                )

                # CG coefficient of the final state
                C_final = cg_coefficient(
                    j1, m1_final,
                    j2, m2_final,
                    Stotal, M
                )

                coefficient = C_initial * C_final

                # Energy eigenvalue
                E_S = lam * Stotal * (Stotal + 1)

                # Evolution e^{-i E_S t}
                real_part += coefficient * np.cos(E_S * t)
                imag_part -= coefficient * np.sin(E_S * t)

            # Transition probability
            P_ij = real_part**2 + imag_part**2

            # Since the weight of the i=N2 sector is exactly 1,
            # directly add j * P_ij.
            Nmu += j * P_ij

        # Probability for one neutrino in beam 1
        return Nmu / N1

    # ============================================================
    # GENERAL CASE: alpha != pi/2
    # ============================================================

    for i in range(1, N2 + 1):

        # --------------------------------------------------------
        # Total magnetic quantum number
        # --------------------------------------------------------

        M = Jmax - i

        # --------------------------------------------------------
        # Weight of the i-th sector
        #
        # C(N2, i)
        # * (cos^2 alpha)^(N2-i)
        # * (sin^2 alpha)^i
        #
        # Calculate using logarithms for numerical stability.
        # --------------------------------------------------------

        if c != 0:
            log_weight = (
                gammaln(N2 + 1)
                - gammaln(i + 1)
                - gammaln(N2 - i + 1)
                + 2 * (N2 - i) * np.log(abs(c))
            )
        else:
            # If cos(alpha)=0, this sector has zero weight
            # for every i < N2.
            log_weight = -np.inf

        if s != 0:
            log_weight += 2 * i * np.log(abs(s))
        elif i > 0:
            log_weight = -np.inf

        weight = np.exp(log_weight)

        # --------------------------------------------------------
        # Allowed total-spin range
        #
        # S = Jmax, Jmax-1, ..., Smin
        #
        # Smin = max(Jmax-i, |j1-j2|)
        # --------------------------------------------------------

        Smin = max(M, abs(j1 - j2))

        # --------------------------------------------------------
        # j = number of nu_e -> nu_mu conversions
        # --------------------------------------------------------

        for j in range(1, min(i, N1) + 1):

            # Initial product state
            m1_initial = j1
            m2_initial = j2 - i

            # Final product state
            m1_final = j1 - j
            m2_final = j2 - i + j

            # Real and imaginary parts of transition amplitude
            real_part = np.zeros_like(t, dtype=float)
            imag_part = np.zeros_like(t, dtype=float)

            # ----------------------------------------------------
            # Sum over total spin S
            # ----------------------------------------------------

            S_values = np.arange(Smin, Jmax + 1, 1)

            for Stotal in S_values:

                # CG coefficient for initial product state
                C_initial = cg_coefficient(
                    j1, m1_initial,
                    j2, m2_initial,
                    Stotal, M
                )

                # CG coefficient for final product state
                C_final = cg_coefficient(
                    j1, m1_final,
                    j2, m2_final,
                    Stotal, M
                )

                coefficient = C_initial * C_final

                # Energy eigenvalue
                E_S = lam * Stotal * (Stotal + 1)

                # Evolution e^{-i E_S t}
                real_part += coefficient * np.cos(E_S * t)
                imag_part -= coefficient * np.sin(E_S * t)

            # Transition probability
            P_ij = real_part**2 + imag_part**2

            # Add weighted contribution to expected number
            # of muon neutrinos in beam 1
            Nmu += weight * j * P_ij

    # ------------------------------------------------------------
    # Convert expected number of muon neutrinos to the
    # conversion probability for one neutrino in beam 1
    # ------------------------------------------------------------

    P_emu = Nmu / N1

    return P_emu
```
