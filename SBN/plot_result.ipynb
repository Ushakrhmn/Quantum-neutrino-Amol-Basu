```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 3

loc = ['results/4nu_ideal', 'results/4nu_swap_approx995', 'results/noswap_approx995']
names = ['Ideal', 'Swap', 'No Swap']

savename = 'swapcompare'

T, P, Er, Sig = [], [], [], []
for i in range(N):
    T.append(np.load(f'{loc[i]}_t.npy'))
    P.append(np.load(f'{loc[i]}_p.npy'))
    Er.append(np.load(f'{loc[i]}_error_rate.npy'))
    Sig.append(np.load(f'{loc[i]}_staterror.npy'))
    print(f'Loaded {loc[i]}')
```

```python
for i in range(N):
    if i == 0:
        plt.plot(T[i], P[i], label=names[i], color='black', linewidth=1, linestyle='--')
    else:
        plt.errorbar(T[i], P[i], yerr=Sig[i], label=names[i], fmt='x', markersize=7.5)

plt.xlabel(r'Time ($1/\mu$)')
plt.ylabel('survival probability')
plt.legend()

plt.savefig(savename+".png")
```

```python
for i in range(N):
    plt.scatter(T[i], Er[i], label=names[i], s=1)
plt.xlabel(r'Time ($1/\mu$)')
plt.ylabel('Error rate')

plt.legend()
plt.savefig(savename+"_error_estimate.png")
```
