```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 1

loc = ['results/test']
names = ['test']

savename = 'test'

T, P, Er, Sig = [], [], [], []
for _ in range(N):
    T.append(np.load(f'{loc[0]}_t.npy'))
    P.append(np.load(f'{loc[0]}_p.npy'))
    Er.append(np.load(f'{loc[0]}_error_rate.npy'))
    Sig.append(np.load(f'{loc[0]}_staterror.npy'))
    print(f'Loaded {loc[0]}')
```

```python
for i in range(N):
    plt.errorbar(T[i], P[i], yerr=Sig[i], label=names[i], fmt='x', markersize=3)

plt.xlabel(r'Time ($1/\mu$)')
plt.ylabel('survival probability')
plt.legend()

plt.savefig(savename+".png")
```

```python
for i in range(N):
    plt.scatter(T[i], Er[i], label=names[i], s=1)
plt.xlabel(r'Time ($1/\mu$)')
plt.ylabel('Error rate')

plt.legend()
plt.savefig(savename+"_error_estimate.png")
```
