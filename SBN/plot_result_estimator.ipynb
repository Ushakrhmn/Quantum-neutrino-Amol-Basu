```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 2

loc = ['results/8nu_estimator_Ideal', 'results/8nu_torino_estimator_res2']
names = ["Ideal", "Torino"]

savename = '8nu_torino_estimator'


I_t = np.load(f'{loc[0]}_t.npy')

Torino_t = np.load(f'{loc[1]}_t.npy')

I_evs = np.load(f'{loc[0]}_evs.npy')
Torino_evs = np.load(f'{loc[1]}_evs.npy')
print("Loaded.")
```

```python
print(I_evs)
```

```python
for i in range(8): 
   # Plot
   plt.clf()
   plt.plot(I_t, I_evs[:, 3*i], marker='o', linestyle='--', color='blue')
   plt.plot(I_t, I_evs[:, 3*i+1], marker='o', linestyle='--', color='red')
   plt.plot(I_t, I_evs[:, 3*i+2], marker='o', linestyle='--', color='green')

   plt.plot(Torino_t, Torino_evs[:, 3*i], marker='x', linestyle='-', color='blue')
   plt.plot(Torino_t, Torino_evs[:, 3*i+1], marker='x', linestyle='-', color='red')
   plt.plot(Torino_t, Torino_evs[:, 3*i+2], marker='x', linestyle='-', color='green')

   custom_lines = [
      plt.Line2D([0], [0], color='blue', lw=2, label=r'$\nu_e$'),
      plt.Line2D([0], [0], color='red', lw=2, label=r'$\nu_\mu$'),
      plt.Line2D([0], [0], color='green', lw=2, label=r'$\nu_\tau$'),
      plt.Line2D([0], [0], marker='o', color='b', label='Ideal', markerfacecolor='blue', markersize=10, linestyle='--'),
      plt.Line2D([0], [0], marker='x', color='b', label='Torino', markerfacecolor='blue', markersize=10, linestyle='-')
   ]

   # Add the custom legend
   plt.legend(handles=custom_lines, loc='upper right')

   # Show the plot
   plt.xlabel(r"Time ($\mu^{-1}$)")
   plt.ylabel(r"P($\nu_0$ → $\nu_i$)")
   plt.title(r"$\nu_{0}$ Survival Probability".format(i))
   plt.grid(True)
   plt.show()
   # plt.savefig(f"{run_name}_survprob.png")
```
