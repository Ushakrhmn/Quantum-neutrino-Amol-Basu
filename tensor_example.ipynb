```python
import quimb as qu
import matplotlib.pyplot as plt
```

```python
N = 32
```

```python
qc = qu.tensor.Circuit(N)
```

```python
# just applly a H gate on one of the qubits
qc.apply_gate("H", 12)
```

```python
samples = []
for s in  qc.sample(512, seed = 42):
    samples.append(s)
```

```python
plt.hist(samples)
```
