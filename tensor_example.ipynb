```python
# This section shows how to convert a qiskit circuit to quimb
```

```python
qc = QuantumCircuit(2)
qc.x(0)
qc.cz(0, 1)
qc.rz(0.5, 1)
qc.sx(0)
```

```python
circ = qiskit_to_quimb(qc)
```

```python
for g in circ.gates:
    print(g)
```

```python
samples = []
for s in circ.sample(512, seed = 42):
    samples.append(s)
```

```python
plt.hist(samples)
```
