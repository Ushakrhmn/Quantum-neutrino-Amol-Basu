```python
from qiskit_aer import AerSimulator
from JobResult import JobResult
import qiskit as qk
```

```python
simulator = AerSimulator()
```

```python
qc = qk.QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

qc.draw('mpl')
```

```python
job = JobResult(simulator)
```

```python
job.run(qc, {"shots": 1024})
```

```python
job.check_job_status()
```

```python
result = job.get_result()
```
